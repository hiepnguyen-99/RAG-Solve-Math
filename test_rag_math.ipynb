```python
import os
import json
from model.rag_4b import solve_question_4b,solve_math_question_4b
```

```python
with open('data/cleaned/test/dai_so.json', 'r', encoding='utf-8') as f:
    data = json.load(f)
test_data_math = data["problems"]
print(f"Số lượng bài test: {len(test_data_math)}")
```

```python
def solve_math_batch(questions):
    """
    Nhận list câu hỏi, trả về list đáp án bằng cách gọi solve_math_question_4b từng câu.
    Nếu gặp lỗi, trả về None cho câu đó.
    """
    answers = []
    for q in questions:
        try:
            result = solve_math_question_4b(q)
            ans = result if isinstance(result, str) else result[0]
        except Exception as e:
            ans = None
        answers.append(ans)
    return answers
```

```python

def process_in_batches(questions, batch_size):
    """
    """
    all_answers = []
    for i in range(0, len(questions), batch_size):
        batch = questions[i:i+batch_size]
        batch_answers = solve_math_batch(batch)
        all_answers.extend(batch_answers)
    return all_answers
```

TEST giải toán

```python
# Test giải toán trên dữ liệu test và ghi kết quả theo batch_size
batch_size = 32  
questions = [item['question'] for item in test_data_math]
answers = process_in_batches(questions, batch_size=batch_size)
results_math = []
for idx, (question, answer) in enumerate(zip(questions, answers), 1):
    results_math.append({"index": idx, "input": question, "output": answer})
os.makedirs('Answer', exist_ok=True)

with open('Answer/answer_math.json', 'w', encoding='utf-8') as f:
    json.dump(results_math, f, ensure_ascii=False, indent=2)
print(f"Đã ghi {len(results_math)} kết quả ra file Answer/answer_math.json")
```
